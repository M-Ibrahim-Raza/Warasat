### __*Function for Capitalizing Words for Display*__

```python
def capitalizeWords(s):
    return " ".join(word.capitalize() for word in s.split(" "))
```

### __*Function for Formatting Numbers for Display*__

```python
def formatNumber(num):
    if num > 1000:
        return int(num) if num.is_integer() else round(num)  # Return integer if no decimal places
    elif num > 100:
        return int(num) if num.is_integer() else f"{num:.1f}"  # Show 1 decimal place if needed
    else:
        return int(num) if num.is_integer() else f"{num:.2f}"  # Show 2 decimal places if needed
```

### __*Function for Calculating Percentage*__

```python
def calculatePercentage(value, total):
    if total == 0:
        return "0%"  # Avoid division by zero
    percentage = (value / total) * 100
    return f"{percentage:.0f}%" if percentage.is_integer() else f"{percentage:.2f}%"
```

### __*Function for Calculating Value of Percentage*__

```python
def calculatePercentageValue(percentage, total):
    if total == 0:
        return 0  # Avoid division by zero
    value = (percentage * total) / 100
    return value
```

### __*Function for Adding General Details of Inheritance Calculations*__

```python
def AddGeneralDetails(ws,name_of_deceased_person="",date_of_death="",gender=""):

    # Inputing Values into Specified Cells
    ws["B1"] = current_date  # Calculation Date (Current Date)
    ws["B2"] = name_of_deceased_person  # Name of Deceased Person
    ws["B3"] = date_of_death  # Date of Death
    ws["B4"] = gender  # Gender
```

### __*Function for Adding Asset Details Table*__

```python
def addAssetDetails(ws, asset_details, currency=""):

    # Appending Distributable Amount to Asset Details
    asset_details["Distributable Amount"]=0
    
    # Merging A6 and B6 for "Asset Details" Heading
    ws.merge_cells("A6:B6")
    ws["A6"] = "Asset Details"
    
    # Defining Border Style
    thin_border = Border(
        left=Side(style="thin", color="000000"),
        right=Side(style="thin", color="000000"),
        top=Side(style="thin", color="000000"),
        bottom=Side(style="thin", color="000000"),
    )

    # Applying Styling to Heading
    ws["A6"].font = Font(bold=True, size=14, name="Times New Roman")  
    ws["A6"].alignment = Alignment(horizontal="center", vertical="center")
    ws["A6"].border = thin_border  # Apply border to heading
    ws["B6"].border = thin_border  # Apply border to merged cell

    # Inserting Data From Row 7
    start_row = 7
    last_row = start_row + len(asset_details) # Calculate the Last Row Index
    distributable_row = None  # To Store the Row Number of "Distributable Amount"

    for row_idx, (key, value) in enumerate(asset_details.items(), start=start_row):
        cell_key = ws[f"A{row_idx}"]
        cell_value = ws[f"B{row_idx}"]

        # Insert Key
        cell_key.value = key

        if key == "Distributable Amount":
            # Storing Row Index for Distributable Amount
            distributable_row = row_idx

            # Applying Formula : Total Amount - SUM(other amounts)
            if len(asset_details)==2:
                formula=f"=B{start_row}"
            else:
                formula = f"=B{start_row} - SUM(B{start_row+1}:B{row_idx-1})"
            cell_value.value = formula
        else:
            # Inserting Numeric Value for other Rows
            cell_value.value = value  

        # Applying Styles
        cell_key.font = Font(bold=True, name="Times New Roman")  # Bold for Keys
        cell_value.font = Font(name="Times New Roman")
        cell_key.alignment = Alignment(horizontal="center", vertical="center")
        cell_value.alignment = Alignment(horizontal="center", vertical="center")

        if key == "Total Amount":
            cell_value.font = Font(bold=True, name="Times New Roman")  # Bold for "Total Amount"
        if key == "Distributable Amount":
            cell_key.font = Font(bold=True, color="FF0000", name="Times New Roman")  # Red & Bold for Key
            cell_value.font = Font(bold=True, color="FF0000", name="Times New Roman")  # Red & Bold for Value

        # Applying Borders
        cell_key.border = thin_border
        cell_value.border = thin_border

        # Applying Excel Custom Formatting for Currency
        cell_value.number_format = f'"{currency} "#,##0'
```

### __*Function for Adding Heir Details Table*__

```python
def addHeirDetails(ws, heir_details, distributable_amount_cell, start, currency=""):

    headers = ["Count", "Heir Relation", "Heir Category", "Shares", "Shares Amount", "Share For Each Person", "Share Amount For Each Person"]
    col_letters = ["A", "B", "C", "D", "E", "F", "G"]

    # Defining Border Style
    thin_border = Border(
        left=Side(style="thin", color="000000"),
        right=Side(style="thin", color="000000"),
        top=Side(style="thin", color="000000"),
        bottom=Side(style="thin", color="000000"),
    )
    
    # Merging A{start} to G{start} for "Heir Details" Heading
    ws.merge_cells(f"A{start}:G{start}")
    ws[f"A{start}"] = "Heir Details"
    
    # Styling Heading
    ws[f"A{start}"].font = Font(bold=True, size=14, name="Times New Roman")  
    ws[f"A{start}"].alignment = Alignment(horizontal="center", vertical="center")
    for col in col_letters:
        ws[f"{col}{start}"].border = thin_border

    # Inserting Table Header Row
    header_row = start + 1  # The row index for the headers

    for col, header in zip(col_letters, headers):
        cell = ws[f"{col}{header_row}"]
        cell.value = header
        cell.font = Font(bold=True, name="Times New Roman")
        cell.alignment = Alignment(horizontal="center", vertical="center")
        cell.border = thin_border  # Apply border

    # Inserting Data for Heir Details
    for row_idx, heir in enumerate(heir_details, start=header_row + 1):

        count_cell = ws[f"A{row_idx}"]
        heir_relation_cell = ws[f"B{row_idx}"]
        heir_category_cell = ws[f"C{row_idx}"]
        shares_cell = ws[f"D{row_idx}"]
        shares_amount_cell = ws[f"E{row_idx}"]
        shares_per_person_cell = ws[f"F{row_idx}"]
        shares_per_person_amount_cell = ws[f"G{row_idx}"]

        count_cell.value = heir["Count"]
        heir_relation_cell.value = heir["Heir Relation"]
        heir_category_cell.value = heir["Category"]
        shares_cell.value = heir["Shares"]
        shares_amount_cell.value = f"=D{row_idx}*{distributable_amount_cell}"
        shares_per_person_cell.value = f"=D{row_idx}/A{row_idx}"
        shares_per_person_amount_cell.value = f"=F{row_idx}*{distributable_amount_cell}"

        # Applying Styles
        for cell in [count_cell, heir_relation_cell, heir_category_cell, shares_cell, shares_amount_cell, shares_per_person_cell, shares_per_person_amount_cell]:
            cell.font = Font(name="Times New Roman")
            cell.alignment = Alignment(horizontal="center", vertical="center")
            cell.border = thin_border

        # Applying Formatting
        shares_cell.number_format = '0.00%'
        shares_per_person_cell.number_format = '0.00%'
        shares_amount_cell.number_format = f'"{currency} "#,##0'
        shares_per_person_amount_cell.number_format = f'"{currency} "#,##0'

         
    # Auto-Adjusting Column Width
    for col_idx in range(1, ws.max_column + 1):  # Iterate over all columns
        col_letter = get_column_letter(col_idx)  # Convert column index to letter
        max_length = 0  # Track the longest text length in the column

        for row in ws.iter_rows(min_row=1, max_row=ws.max_row, min_col=col_idx, max_col=col_idx):
            for cell in row:
                if cell.value:
                    max_length = max(max_length, len(str(cell.value)))

    # Setting Column Width with Padding
        ws.column_dimensions[col_letter].width = max_length + 3
```

## __*PDF Generation*__

### __*Function for Replacing Placeholders in Word Doc While Preserving Format*__

```python
# Function to replace text while handling split runs
def replaceText(doc, old_text, new_text):
    for para in doc.paragraphs:
        full_text = "".join(run.text for run in para.runs)  # Combine all runs
        if old_text in full_text:
            full_text = full_text.replace(old_text, new_text)  # Replace placeholder
            
            # Clear existing runs
            for run in para.runs:
                run.text = ""

            # Reapply the modified text in the first run (keeps formatting)
            para.runs[0].text = full_text  

    # Also check inside tables (if placeholders are inside tables)
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                for para in cell.paragraphs:
                    full_text = "".join(run.text for run in para.runs)
                    if old_text in full_text:
                        full_text = full_text.replace(old_text, new_text)
                        for run in para.runs:
                            run.text = ""
                        para.runs[0].text = full_text
```

### __*Function to Add Asset Details Table in Word Doc*__

```python
def addAssetDetailsTable(doc,asset_details):
    for para in doc.paragraphs:
        if "Asset Details" in para.text:
            insert_paragraph = para
            break
    
    # Insert a table with 2 columns and (n+1) rows (including header)
    table = doc.add_table(rows=len(asset_details), cols=2)
    
    # Apply formatting to table
    table.style = 'Table Grid'
    
    for row_idx, (key, value) in enumerate(asset_details.items()):
        # Determine styling based on row position
        is_first_row = row_idx == 0
        is_last_row = row_idx == len(asset_details) - 1
    
        # Left column (bold for first & last rows)
        left_cell = table.cell(row_idx, 0)
        left_run = left_cell.paragraphs[0].add_run(key)
        left_run.bold = True
        left_run.font.name = "Times New Roman"
        left_run.font.size = Pt(16 if is_first_row or is_last_row else 14)
    
        # Apply red color for last row
        if is_last_row:
            left_run.font.color.rgb = RGBColor(255, 0, 0)
    
        # Right column (bold for first & last rows)
        right_cell = table.cell(row_idx, 1)
        right_run = right_cell.paragraphs[0].add_run(str(value))
        right_run.bold = True if is_first_row or is_last_row else False
        right_run.font.name = "Times New Roman"
        right_run.font.size = Pt(16 if is_first_row or is_last_row else 14)
    
        # Apply red color for last row
        if is_last_row:
            right_run.font.color.rgb = RGBColor(255, 0, 0)
    
        # Center align text
        for cell in [left_cell, right_cell]:
            cell.paragraphs[0].alignment = 1  # Center alignment
    
    # Move table below "Asset Details"
    insert_paragraph.insert_paragraph_before("\n")  # Add space before table
    insert_paragraph._element.addnext(table._element)  # Insert table after "Asset Details"
```

### __*Function to Set Vertical Alignment of Cell to Center*__

```python
def setVerticalAlignment(cell):
    """Set vertical alignment to center."""
    tc = cell._element
    tcPr = tc.find(qn('w:tcPr')) or OxmlElement('w:tcPr')
    vAlign = OxmlElement('w:vAlign')
    vAlign.set(qn('w:val'), "center")
    tcPr.append(vAlign)
    tc.append(tcPr)
```

### __*Function to Add Heir Details Table in Word Doc*__

```python
def addHeirDetailTable(doc, heir_details):
    for para in doc.paragraphs:
        if "Heir Details" in para.text:
            heir_paragraph = para
            break
    
    # Define table headers
    headers = ["Count", "Heir Relation", "Category", "Shares", "Shares Amount",
               "Share for Each Person", "Share Amount for Each Person"]
    
    # Insert table with headers + rows for heir details
    heir_table = doc.add_table(rows=1 + len(heir_details), cols=len(headers))
    heir_table.style = 'Table Grid'
    
    # Define column widths
    narrow_cols = [0, 1, 2, 3]  # First 4 columns (narrower)
    wide_cols = [5, 6]  # Last 2 columns (wider)
    
    # Format header row
    for col_idx, header in enumerate(headers):
        cell = heir_table.cell(0, col_idx)
        run = cell.paragraphs[0].add_run(header)
        run.bold = True
        run.font.name = "Times New Roman"
        run.font.size = Pt(14)
        cell.paragraphs[0].alignment = WD_PARAGRAPH_ALIGNMENT.CENTER  # Horizontal center
        setVerticalAlignment(cell)  # Vertical center
        
        # Adjust column widths
        if col_idx in narrow_cols:
            cell.width = Inches(1.0)  # Narrower
        elif col_idx in wide_cols:
            cell.width = Inches(1.8)  # Wider
    
    # Fill table rows
    for row_idx, heir in enumerate(heir_details, start=1):
        for col_idx, key in enumerate(headers):
            cell = heir_table.cell(row_idx, col_idx)
            run = cell.paragraphs[0].add_run(str(heir[key]))
            run.font.name = "Times New Roman"
            run.font.size = Pt(14)
            cell.paragraphs[0].alignment = WD_PARAGRAPH_ALIGNMENT.CENTER  # Horizontal center
            setVerticalAlignment(cell)  # Vertical center
            
            # Adjust column widths for content rows
            if col_idx in narrow_cols:
                cell.width = Inches(1.0)  # Narrower
            elif col_idx in wide_cols:
                cell.width = Inches(1.8)  # Wider
    
    # Move table below "Heir Details"
    heir_paragraph.insert_paragraph_before("\n")  # Add space before table
    heir_paragraph._element.addnext(heir_table._element)  # Insert table after "Heir Details"
```

## __*Inheritance Calculation*__

### __*Mapping Dictionaries*__

```python
relation_map = {
    "son": "num_sons",
    "daughter": "num_daughters",
    "father": "has_father",
    "mother": "has_mother",
    "husband": "has_husband",
    "wife": "num_wives",
    "son's son": "num_sons_sons",
    "son's daughter": "num_sons_daughters",
    "father's father": "has_fathers_father",
    "father's mother": "has_fathers_mother",
    "mother's mother": "has_mothers_mother",
    "real brother": "num_real_brothers",
    "real sister": "num_real_sisters",
    "paternal brother": "num_paternal_brothers",
    "paternal sister": "num_paternal_sisters",
    "maternal sibling": "num_maternal_siblings",
    "son's son's son": "num_sons_sons_sons",
    "son's son's daughter": "num_sons_sons_daughters"
}

share_map = {
    "HusbandShare": "husband",
    "PerWifeShare": "wife",
    "PerSonShare": "son",
    "PerDaughterShare": "daughter",
    "PerSonsDaughterShare": "son's daughter",
    "FatherShare": "father",
    "PerSonsSonsDaughterShare": "son's son's daughter",
    "MotherShare": "mother",
    "FathersFatherShare": "father's father",
    "FathersMotherShare": "father's mother",
    "MothersMotherShare": "mother's mother",
    "PerRealSisterShare": "real sister",
    "PerPaternalSisterShare": "paternal sister",
    "PerMaternalSiblingShare": "maternal sibling",
    "PerSonsSonShare": "son's son",
    "PerSonsSonsSonShare": "son's son's son",
    "PerRealBrotherShare": "real brother",
    "PerPaternalBrotherShare": "paternal brother"
}
```

### __*Dictionary for Relation Counts*__

```python
relation_counts = {
    "num_wives": 0,
    "num_sons": 0,
    "num_daughters": 0,
    "num_sons_sons": 0,
    "num_sons_daughters": 0,
    "has_father": 0,
    "num_sons_sons_daughters": 0,
    "num_sons_sons_sons": 0,
    "has_mother": 0,
    "num_real_brothers": 0,
    "num_real_sisters": 0,
    "num_paternal_brothers": 0,
    "num_paternal_sisters": 0,
    "num_maternal_siblings": 0,
    "has_fathers_father": 0,
    "has_fathers_mother": 0,
    "has_mothers_mother": 0,
    "has_husband": 0
}
```

### __*Function for Relation Counts Calcuation*__

```python
def relationCountsCalculation(heir_list):
    temp_relation_counts = relation_counts
    for heir in heir_list:
        relation = heir.get("relation")
        val = heir.get("val", 0)
        temp_relation_counts[relation_map[relation]] = val
    return temp_relation_counts
```

### __*Function for Assigning Shares to Each Heir*__

```python
def assignHeirShares(shares,heir_list):
    for heir in heir_list:
        relation = heir.get("relation")
        heir["amount"] = shares.get(next((key for key, value in share_map.items() if value == relation), ""), 0)
```

## __*API Routes*__

```python
# Creating Flask Server App
app = Flask(__name__)

# Enables Cross Origin Resources Sharing in Flask App
CORS(app)

# Load Prolog Rules from the rule.pl file
prolog = Prolog()
prolog.consult("rules.pl")

# -------------------------------------------------------------------------- INHERITANCE CALCULATION API -------------------------------------------------------------------------- 
# Defining API Route for Inheritance Calculation
@app.route('/inheritance-calculator-2', methods=['POST'])
def inheritance_calculator_2():

    # Extracting Received Data
    data = request.get_json()

    # Exceptional Handling
    if not data:
        return jsonify({"error": "Invalid JSON input"}), 400
    
    total_wealth = data.get('total_amount', 0)  # Extracting Total Amount
    heir_list = data.get('heir_list', [])  # Extracting Heir List

    # Testing
    print("--------------------------------------| RECEIVED INHERITANCE CALCULATION REQUEST FROM CLIENT |--------------------------------------")
    print("Received Data :",)
    print("Total Wealth : ",total_wealth)
    print("Hier List :")
    for heir in heir_list:
        print(heir)

    # Calculating Relation Counts
    relation_counts = relationCountsCalculation(heir_list)
    
    # Querying Prolog for Inheritance Shares
    result = list(prolog.query(
        f"inheritance_calculator({total_wealth}, {relation_counts['has_husband']}, {relation_counts['num_wives']}, "
        f"{relation_counts['num_sons']}, {relation_counts['num_daughters']}, {relation_counts['num_sons_sons']}, "
        f"{relation_counts['num_sons_daughters']}, {relation_counts['num_sons_sons_daughters']}, {relation_counts['num_sons_sons_sons']}, "
        f"{relation_counts['has_father']}, {relation_counts['has_mother']}, {relation_counts['num_real_brothers']}, "
        f"{relation_counts['num_real_sisters']}, {relation_counts['num_paternal_brothers']}, {relation_counts['num_paternal_sisters']}, "
        f"{relation_counts['num_maternal_siblings']}, {relation_counts['has_fathers_father']}, {relation_counts['has_fathers_mother']}, "
        f"{relation_counts['has_mothers_mother']}, HusbandShare, PerWifeShare, PerSonShare, PerDaughterShare, PerSonsDaughterShare, "
        f"FatherShare, PerSonsSonsDaughterShare, MotherShare, FathersFatherShare, FathersMotherShare, MothersMotherShare, "
        f"PerRealSisterShare, PerPaternalSisterShare, PerMaternalSiblingShare, PerSonsSonShare, PerSonsSonsSonShare, "
        f"PerRealBrotherShare, PerPaternalBrotherShare)"
    ))

    # Assigning Shares to Heirs
    if result:
        shares = result[0]
        assignHeirShares(shares,heir_list)

    # Testing
    print()
    print("Response Data :")     
    for heir in heir_list:
        print(heir)
        
    return jsonify({
        "status": "success",
        "heir_list": heir_list,
    })
    
# -------------------------------------------------------------------------- PDF GENERATION API -------------------------------------------------------------------------- 
@app.route('/inheritance-calculation-pdf', methods=['POST'])
def inheritance_calculation_pdf():
    data = request.get_json()

    if not data:
        return jsonify({"error": "Invalid JSON input"}), 400

    # Extracting values from request data
    amount = float(data.get('total_amount') or 0)
    funeral_expenses = float(data.get('funeral_expenses') or 0)
    mehr = float(data.get('mehr') or 0)
    debt = float(data.get('debt') or 0)
    will = float(data.get('will') or 0)
    heir_list = data.get('heir_list', [])
    gender = data.get('gender')
    currency = data.get('currency')

    distributable_amount = amount - funeral_expenses - mehr - debt - will

    asset_details = {
        "Total Amount": f"{currency} {formatNumber(amount)}",
        "Funeral & Burial Expenses": f"{currency} {formatNumber(funeral_expenses)}",
        "Haq Mehr": f"{currency} {formatNumber(mehr)}",
        "Debt & Loans": f"{currency} {formatNumber(debt)}",
        "Will": f"{currency} {formatNumber(will)}",
        "Distributable Amount": f"{currency} {formatNumber(distributable_amount)}"
    }

    # Filtering only values greater than zero
    asset_details = {key: value for key, value in asset_details.items() if value}

    heir_details = [
        {
            "Count": heir.get("val"),
            "Heir Relation": capitalizeWords(heir.get("relation")),
            "Category": capitalizeWords(heir.get("category", ["", ""])[1]),
            "Shares": calculatePercentage(heir.get("amount") * heir.get("val"), distributable_amount),
            "Shares Amount": f"{currency} {formatNumber(heir.get("amount") * heir.get("val"))}",
            "Share for Each Person": calculatePercentage(heir.get("amount"), distributable_amount),
            "Share Amount for Each Person": f"{currency} {formatNumber(heir.get("amount"))}"
        }
        for heir in heir_list
    ]

    # Generating PDF
    template_file = "Inheritance-Calculation-Template.docx"
    modified_word_file = "Modified-Inheritance-Calculation.docx"
    modified_pdf_file = "Modified-Inheritance-Calculation.pdf"

    shutil.copy(template_file, modified_word_file) 

    # Loading Document
    doc = Document(modified_word_file)

    # Replacing values in the document
    replaceText(doc, "<<CalculationDate>>", current_date)
    replaceText(doc, "<<Gender>>", capitalizeWords(gender))
    addAssetDetailsTable(doc, asset_details)
    addHeirDetailTable(doc, heir_details)

    # Save updated document
    doc.save(modified_word_file)

    # Convert Word to PDF
    comtypes.CoInitialize()
    word = CreateObject("Word.Application")
    doc = word.Documents.Open(os.path.abspath(modified_word_file))
    doc.SaveAs(os.path.abspath(modified_pdf_file), FileFormat=17)  # PDF format
    doc.Close()
    word.Quit()
    comtypes.CoUninitialize()

    # Delete Word file
    os.remove(modified_word_file)

    # Testing - Print Request Details
    print("\n--------------------------------------| RECEIVED PDF GENERATION REQUEST FROM CLIENT |--------------------------------------")
    print("Asset Details :")
    for key, value in asset_details.items():
        print(f"{key}: {value}")
    
    print("\nHeir Details :")
    for heir in heir_details:
        print(heir)

    print("\n✅ PDF Generated Successfully!")
  
    # Send the generated PDF file to the client
    return send_file(modified_pdf_file, as_attachment=True)

# -------------------------------------------------------------------------- EXCEL GENERATION API -------------------------------------------------------------------------- 
@app.route('/inheritance-calculation-xlsx', methods=['POST'])
def inheritance_calculation_xlsx():
    data = request.get_json()

    if not data:
        return jsonify({"error": "Invalid JSON input"}), 400

    # Extracting values from request data
    amount = float(data.get('total_amount') or 0)
    funeral_expenses = float(data.get('funeral_expenses') or 0)
    mehr = float(data.get('mehr') or 0)
    debt = float(data.get('debt') or 0)
    will = float(data.get('will') or 0)
    heir_list = data.get('heir_list', [])
    gender = data.get('gender')
    currency = data.get('currency')

    distributable_amount = amount - funeral_expenses - mehr - debt - will

    asset_details = {
        "Total Amount": amount,
        "Funeral & Burial Expenses": funeral_expenses,
        "Haq Mehr": mehr,
        "Debt & Loans": debt,
        "Will": will,
        "Distributable Amount": distributable_amount,  # Adding this to the asset details
    }

    # Filtering only values greater than zero
    asset_details = {key: value for key, value in asset_details.items() if value}

    heir_details = [
        {
            "Count": heir.get("val"),
            "Heir Relation": capitalizeWords(heir.get("relation")),
            "Category": capitalizeWords(heir.get("category", ["", ""])[1]),
            "Shares": heir.get('amount') * heir.get("val") / distributable_amount,
        }
        for heir in heir_list
    ]

    # Generating Excel File
    excel_template_file = "Inheritance-Calculation-Template.xlsx"
    excel_new_file = "Inheritance-Calculation.xlsx"
    
    shutil.copy(excel_template_file, excel_new_file)

    # Loading Excel File
    wb = openpyxl.load_workbook(excel_new_file)
    ws = wb.active  

    # Adding Data to Excel
    AddGeneralDetails(ws=ws, gender=capitalizeWords(gender))
    addAssetDetails(ws, asset_details, currency="Rs")
    addHeirDetails(ws, heir_details, start=len(asset_details)+8, distributable_amount_cell=f"B{len(asset_details)+6}", currency="Rs")

    # Save Excel File
    wb.save(excel_new_file)
    wb.close()

    # Send File as Response
    return send_file(excel_new_file, as_attachment=True, mimetype="application/vnd.openxmlformats-officedocument.spreadsheetml.sheet")
    
if __name__ == '__main__':
    app.run(debug=False, port=8080)
```
