```python
from flask import Flask, request, jsonify
from pyswip import Prolog
app = Flask(__name__)
prolog = Prolog()
prolog.consult("rules.pl")

@app.route('/inheritance-calculator', methods=['POST'])
def inheritance_calculator():
    try:
        # Get JSON data from the request
        data = request.get_json()
        total_wealth = data['total_wealth']
        has_husband = data.get('has_husband', 0)
        num_wives = data.get('num_wives', 0)
        num_sons = data.get('num_sons', 0)
        num_daughters = data.get('num_daughters' ,0)
        num_sons_sons = data.get('num_sons_sons' ,0)
        num_sons_daughters = data.get('num_sons_daughters' ,0)
        has_father = data.get('has_father',0)
        num_sons_sons_daughters = data.get('num_sons_sons_daughters' ,0)
        num_sons_sons_sons = data.get('num_sons_sons_sons' ,0)
        has_mother = data.get('has_mother',0)
        num_real_brothers = data.get('num_real_brothers',0)
        num_real_sisters = data.get('num_real_sisters',0)
        num_paternal_brothers = data.get('num_paternal_brothers',0)
        num_paternal_sisters = data.get('num_paternal_sisters',0)
        num_maternal_siblings = data.get('num_maternal_siblings',0)
        has_fathers_father = data.get('has_fathers_father',0)
        has_fathers_mother = data.get('has_fathers_mother',0)
        has_mothers_mother = data.get('has_mothers_mother',0)

        # Validate input constraints
        if num_wives > 4:
            return jsonify({"error": "Number of wives cannot exceed 4"}), 400
        if has_husband == 1 and num_wives != 0:
            return jsonify({"error": "If husband is present, number of wives must be 0"}), 400

        # Query the Prolog engine
        result = list(prolog.query(
    f"inheritance_calculator({total_wealth}, {has_husband}, {num_wives}, {num_sons}, {num_daughters}, {num_sons_sons}, {num_sons_daughters}, {num_sons_sons_daughters}, "
    f"{num_sons_sons_sons}, {has_father}, {has_mother}, {num_real_brothers}, {num_real_sisters}, {num_paternal_brothers}, {num_paternal_sisters}, {num_maternal_siblings}, "
    f"{has_fathers_father}, {has_fathers_mother}, {has_mothers_mother}, HusbandShare, PerWifeShare, PerSonShare, PerDaughterShare, PerSonsDaughterShare, FatherShare, "
    f"PerSonsSonsDaughterShare, MotherShare, FathersFatherShare, FathersMotherShare, MothersMotherShare, PerRealSisterShare, PerPaternalSisterShare, PerMaternalSiblingShare, "
    f"PerSonsSonShare, PerSonsSonsSonShare, PerRealBrotherShare, PerPaternalBrotherShare)"
))

    
        # Process the result
        if result:
            response = result[0]
            return jsonify({
                "husband_share": response['HusbandShare'],
                "per_wife_share": response['PerWifeShare'],
                "per_son_share": response['PerSonShare'],
                "per_daughter_share": response['PerDaughterShare'],
                "per_sons_daughter_share": response['PerSonsDaughterShare'],
                "father_share": response['FatherShare'],
                "per_sons_sons_daughter_share": response['PerSonsSonsDaughterShare'],
                "mother_share": response['MotherShare'],
                "fathers_father_share": response['FathersFatherShare'],
                "fathers_mother_share": response['FathersMotherShare'],
                "mothers_mother_share": response['MothersMotherShare'],
                "per_real_sister_share":response['PerRealSisterShare'],
                "per_paternal_sister_share":response['PerPaternalSisterShare'],
                "per_maternal_sibling_share":response['PerMaternalSiblingShare'],
                "per_sons_son_share":response['PerSonsSonShare'],
                "per_sons_sons_son_share":response['PerSonsSonsSonShare'],
                "per_real_brother_share":response['PerRealBrotherShare'],
                "per_paternal_brother_share":response['PerPaternalBrotherShare']
            })
        else:
            return jsonify({"error": "Calculation failed"}), 400
    except KeyError as e:
        return jsonify({"error": f"Missing required field: {str(e)}"}), 400
    except Exception as e:
        return jsonify({"error": str(e)}), 500


if __name__ == '__main__':
    app.run(debug=False)
```

```python
from flask import Flask, request, jsonify
from flask_cors import CORS
from pyswip import Prolog
app = Flask(__name__)
CORS(app)
prolog = Prolog()
prolog.consult("rules.pl")
@app.route('/inheritance-calculator-2', methods=['POST'])
def inheritance_calculator_2():
    data = request.get_json()
    print("Received Data:", data)

    heirs = data.get('heirs', [])  # Extract heirs list, default to empty list
    total_wealth = data.get('total_amount', 0)  # Get total amount
    print(total_wealth)

    # Dictionary to count relations
    relation_counts = {
        "num_wives": 0,
        "num_sons": 0,
        "num_daughters": 0,
        "num_sons_sons": 0,
        "num_sons_daughters": 0,
        "has_father": 0,
        "num_sons_sons_daughters": 0,
        "num_sons_sons_sons": 0,
        "has_mother": 0,
        "num_real_brothers": 0,
        "num_real_sisters": 0,
        "num_paternal_brothers": 0,
        "num_paternal_sisters": 0,
        "num_maternal_siblings": 0,
        "has_fathers_father": 0,
        "has_fathers_mother": 0,
        "has_mothers_mother": 0,
        "has_husband":0
    }

    # Mapping relations from received data to predefined keys
    relation_map = {
        "son": "num_sons",
        "daughter": "num_daughters",
        "father": "has_father",
        "mother": "has_mother",
        "husband": "has_husband",
        "wife": "num_wives",
        "son's son": "num_sons_sons",
        "son's daughter": "num_sons_daughters",
        "father's father": "has_fathers_father",
        "father's mother": "has_fathers_mother",
        "mother's mother": "has_mothers_mother",
        "real brother": "num_real_brothers",
        "real sister": "num_real_sisters",
        "paternal brother": "num_paternal_brothers",
        "paternal sister": "num_paternal_sisters",
        "maternal sibling": "num_maternal_siblings",
        "son's son's son": "num_sons_sons_sons",
        "son's son's daughter": "num_sons_sons_daughters",
        "father's father's father": "has_fathers_fathers_father"
    }


    # Count relations based on received data
    for heir in heirs:
        relation = heir.get("relation")
        val = heir.get("val", 0)  # Extract 'val', default to 0

        if relation in relation_map:
            relation_counts[relation_map[relation]] += val  # Add val count

    # Printing counts for debugging
    print("Extracted Counts:", relation_counts)

    has_husband = relation_counts["has_husband"]
    num_wives = relation_counts["num_wives"]
    num_sons = relation_counts["num_sons"]
    num_daughters = relation_counts["num_daughters"]
    num_sons_sons = relation_counts["num_sons_sons"]
    num_sons_daughters = relation_counts["num_sons_daughters"]
    num_sons_sons_daughters = relation_counts["num_sons_sons_daughters"]
    num_sons_sons_sons = relation_counts["num_sons_sons_sons"]
    has_father = relation_counts["has_father"]
    has_mother = relation_counts["has_mother"]
    num_real_brothers = relation_counts["num_real_brothers"]
    num_real_sisters = relation_counts["num_real_sisters"]
    num_paternal_brothers = relation_counts["num_paternal_brothers"]
    num_paternal_sisters = relation_counts["num_paternal_sisters"]
    num_maternal_siblings = relation_counts["num_maternal_siblings"]
    has_fathers_father = relation_counts["has_fathers_father"]
    has_fathers_mother = relation_counts["has_fathers_mother"]
    has_mothers_mother = relation_counts["has_mothers_mother"]

    # Query Prolog for inheritance distribution
    result = list(prolog.query(
        f"inheritance_calculator({total_wealth}, {has_husband}, {num_wives}, {num_sons}, {num_daughters}, {num_sons_sons}, {num_sons_daughters}, {num_sons_sons_daughters}, "
        f"{num_sons_sons_sons}, {has_father}, {has_mother}, {num_real_brothers}, {num_real_sisters}, {num_paternal_brothers}, {num_paternal_sisters}, {num_maternal_siblings}, "
        f"{has_fathers_father}, {has_fathers_mother}, {has_mothers_mother}, HusbandShare, PerWifeShare, PerSonShare, PerDaughterShare, PerSonsDaughterShare, FatherShare, "
        f"PerSonsSonsDaughterShare, MotherShare, FathersFatherShare, FathersMotherShare, MothersMotherShare, PerRealSisterShare, PerPaternalSisterShare, PerMaternalSiblingShare, "
        f"PerSonsSonShare, PerSonsSonsSonShare, PerRealBrotherShare, PerPaternalBrotherShare)"
    ))

    if result:
        relation_shares = [
            "HusbandShare", "PerWifeShare", "PerSonShare", "PerDaughterShare",
            "PerSonsDaughterShare", "FatherShare", "PerSonsSonsDaughterShare",
            "MotherShare", "FathersFatherShare", "FathersMotherShare",
            "MothersMotherShare", "PerRealSisterShare", "PerPaternalSisterShare",
            "PerMaternalSiblingShare", "PerSonsSonShare", "PerSonsSonsSonShare",
            "PerRealBrotherShare", "PerPaternalBrotherShare"
        ]

        for relation in relation_shares:
            share = result[0].get(relation, 0)
            if share>0:
                print(f"{relation}: {share}")
    else:
        print("No result from Prolog.")


    
    
    return jsonify({
        "status": "success",
        "extracted_counts": relation_counts,
        "total_amount": total_amount
    })

if __name__ == '__main__':
    app.run(debug=False)
```

```python
from flask import Flask, request, jsonify
from flask_cors import CORS
from pyswip import Prolog

app = Flask(__name__)
CORS(app)
prolog = Prolog()
prolog.consult("rules.pl")

@app.route('/inheritance-calculator-2', methods=['POST'])
def inheritance_calculator_2():
    data = request.get_json()
    print("Received Data:", data)

    heirs = data.get('heirs', [])  # Extract heirs list
    total_wealth = data.get('total_amount', 0)  # Get total amount

    # Dictionary to count relations
    relation_counts = {
        "num_wives": 0,
        "num_sons": 0,
        "num_daughters": 0,
        "num_sons_sons": 0,
        "num_sons_daughters": 0,
        "has_father": 0,
        "num_sons_sons_daughters": 0,
        "num_sons_sons_sons": 0,
        "has_mother": 0,
        "num_real_brothers": 0,
        "num_real_sisters": 0,
        "num_paternal_brothers": 0,
        "num_paternal_sisters": 0,
        "num_maternal_siblings": 0,
        "has_fathers_father": 0,
        "has_fathers_mother": 0,
        "has_mothers_mother": 0,
        "has_husband": 0
    }

    relation_map = {
        "son": "num_sons",
        "daughter": "num_daughters",
        "father": "has_father",
        "mother": "has_mother",
        "husband": "has_husband",
        "wife": "num_wives",
        "son's son": "num_sons_sons",
        "son's daughter": "num_sons_daughters",
        "father's father": "has_fathers_father",
        "father's mother": "has_fathers_mother",
        "mother's mother": "has_mothers_mother",
        "real brother": "num_real_brothers",
        "real sister": "num_real_sisters",
        "paternal brother": "num_paternal_brothers",
        "paternal sister": "num_paternal_sisters",
        "maternal sibling": "num_maternal_siblings",
        "son's son's son": "num_sons_sons_sons",
        "son's son's daughter": "num_sons_sons_daughters"
    }

    for heir in heirs:
        relation = heir.get("relation")
        val = heir.get("val", 0)
        if relation in relation_map:
            relation_counts[relation_map[relation]] += val

    result = list(prolog.query(
        f"inheritance_calculator({total_wealth}, {relation_counts['has_husband']}, {relation_counts['num_wives']}, "
        f"{relation_counts['num_sons']}, {relation_counts['num_daughters']}, {relation_counts['num_sons_sons']}, "
        f"{relation_counts['num_sons_daughters']}, {relation_counts['num_sons_sons_daughters']}, {relation_counts['num_sons_sons_sons']}, "
        f"{relation_counts['has_father']}, {relation_counts['has_mother']}, {relation_counts['num_real_brothers']}, "
        f"{relation_counts['num_real_sisters']}, {relation_counts['num_paternal_brothers']}, {relation_counts['num_paternal_sisters']}, "
        f"{relation_counts['num_maternal_siblings']}, {relation_counts['has_fathers_father']}, {relation_counts['has_fathers_mother']}, "
        f"{relation_counts['has_mothers_mother']}, HusbandShare, PerWifeShare, PerSonShare, PerDaughterShare, PerSonsDaughterShare, "
        f"FatherShare, PerSonsSonsDaughterShare, MotherShare, FathersFatherShare, FathersMotherShare, MothersMotherShare, "
        f"PerRealSisterShare, PerPaternalSisterShare, PerMaternalSiblingShare, PerSonsSonShare, PerSonsSonsSonShare, "
        f"PerRealBrotherShare, PerPaternalBrotherShare)"
    ))

    if result:
        shares = result[0]
        share_map = {
            "HusbandShare": "husband",
            "PerWifeShare": "wife",
            "PerSonShare": "son",
            "PerDaughterShare": "daughter",
            "PerSonsDaughterShare": "son's daughter",
            "FatherShare": "father",
            "PerSonsSonsDaughterShare": "son's son's daughter",
            "MotherShare": "mother",
            "FathersFatherShare": "father's father",
            "FathersMotherShare": "father's mother",
            "MothersMotherShare": "mother's mother",
            "PerRealSisterShare": "real sister",
            "PerPaternalSisterShare": "paternal sister",
            "PerMaternalSiblingShare": "maternal sibling",
            "PerSonsSonShare": "son's son",
            "PerSonsSonsSonShare": "son's son's son",
            "PerRealBrotherShare": "real brother",
            "PerPaternalBrotherShare": "paternal brother"
        }

        for heir in heirs:
            relation = heir.get("relation")
            heir["amount"] = shares.get(next((key for key, value in share_map.items() if value == relation), ""), 0)

    print(heirs)

    return jsonify({
        "status": "success",
        "heirs": heirs,
        "total_amount": total_wealth
    })

if __name__ == '__main__':
    app.run(debug=False)
```
